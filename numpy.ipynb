```python
import numpy as np
#students marks analyzer

marks = np.array([20, 30, 40, 60, 70, 80, 86, 88, 94, 99 ])

print("\nAverage Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))

total_passed = marks[marks >= 50]
total_failed = marks[marks <= 50]

print("\nTotal_Passed:", len(total_passed))
print("Total_Failed:", len(total_failed))
```

```python
#task 5
import numpy as np

marks = np.array([35, 67, 89, 45, 90, 22, 70, 58])

passing_marks = 50

passed = marks[marks >= passing_marks]
failed = marks[marks < passing_marks]

print("Passed marks:", passed)
print("Failed marks:", failed)
print("Number of passed students:", len(passed))
print("Number of failed students:", len(failed))
```
