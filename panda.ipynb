```python
#task 1
#create this student dataframe
import pandas as pd

data = {
    "Name": [
        "Ali",
        "Sara",
        "Ahmed",
        "Ayesha",
        "Zain"
    ],

    "Marks": [
        "70",
        "85",
        "60",
        "95",
        "45"

   ]}

df = pd.DataFrame(data)

# Convert 'Marks' column to numeric type
df['Marks'] = pd.to_numeric(df['Marks'])

print(df)

#for average marks
print("\nAverage Marks:")
print(df["Marks"].mean())

#for maximum marks
print("\nHighest Marks:")
print(df["Marks"].max())

#for minimum marks
print("\nLowest Marks:")
print(df["Marks"].min())

#for total students
total_students = len(df)
print("\nTotal Students:")
print(total_students)
```

```python
#task 2

#students with marks greater than or equal to 50
passed_students = df[df["Marks"] >= 50]
print("\nStudents with Marks >= 50:")
print(passed_students)

#students with marks less than 50
failed_students = df[df["Marks"] < 50]
print("\nStudents with Marks < 50:")
print(failed_students)
```

```python
#task 3
#to add a new column

df["city"] = [
    "Karachi",
    "Lahore",
    "Islamabad",
    "Peshawar",
    "Quetta"
]
df = df[["Name", "Marks", "city"]]
print(df)

#display students only from karachi
karachi_students = df[df["city"] == "Karachi"]
print("\nStudents from Karachi:")
print(karachi_students)
```

```python
#task 4
data = {
    "Name":[
        "Ali",
        "Sara",
        "Ahmed",
        "Ayesha",
        "Zain"
    ],
    "Marks": [
        "70",
        "85",
        "60",
        "95",
        "45"
    ],
    "Course": [
        "AI",
        "DS",
        "AI",
        "DS",
        "AI"
    ] }

df = pd.DataFrame(data)
# Convert 'Marks' column to numeric type
df['Marks'] = pd.to_numeric(df['Marks'])
print(df)

#to find average marks by groupby method
average_marks = df.groupby("Course")["Marks"].mean()
print("\nAverage Marks:")
print(average_marks)
```

```python
#task 5
import pandas as pd
import numpy as np

# Create DataFrame
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [80, np.nan, 70, np.nan, 90]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Find number of missing values
missing_values = df["Marks"].isna().sum()
print("\nNumber of missing values:", missing_values)

# Calculate average marks
average_marks = df["Marks"].mean()
print("Average marks:", average_marks)

# Fill missing values with average marks
df["Marks"] = df["Marks"].fillna(average_marks)

# Display cleaned DataFrame
print("\nCleaned DataFrame:")
print(df)

```
